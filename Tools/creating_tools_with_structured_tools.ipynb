```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class DivideInput(BaseModel):
    a: int = Field( description="the nomenator")
    b: int = Field( description="the denomenator")
```

```python
def mul_func(a, b) -> int:
    return a * b
```

```python
mul_tool = StructuredTool.from_function(
    func=mul_func,
    name='multiply',
    description='Multiply two numbers',
    args_schema=DivideInput
)
```

```python
res = mul_tool.invoke({'a': 3, 'b': 4})
res
```
