```python
from langgraph.graph import StateGraph, START, END 
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
```

```python
model = ChatOpenAI()
```

```python
class Weather(TypedDict):
    user_query: str
    ai_response: str
```

```python
graph = StateGraph(Weather)
```

```python
def weather_func(state: Weather) -> Weather:
    user_query = state['user_query']
    prompt = f'''You are a weather agent and you will only answer the user question asked about weather, 
        if not asked about weather so directly response that i don't know about this and the question is this one {user_query}'''
    ans = model.invoke(prompt).content
    return {'ai_response': ans}
```

```python
graph.add_node("weather_func", weather_func)
graph.add_edge(START, "weather_func")
graph.add_edge("weather_func", END)
```

```python

workflow = graph.compile()
```

```python
# workflow = graph.invoke()
initial_state = {'user_query': 'What is the weather in Peshawar?'}
```

```python
final_state = workflow.invoke(initial_state)
print(final_state['ai_response'])
```
