```python
from langchain_groq import ChatGroq
from langchain.tools import tool
```

```python
@tool
def add(a: int, b: int) -> int:
    """ this tool is provided two inputs and it returns their sum. """
    return a + b

@tool 
def mul(a: int, b: int) -> int:
    """ This tool is provided with two inputs and it returns their product."""
    return a * b
```

```python
print(mul.invoke({'a': 4, 'b': 3}))
print(add.invoke({'a': 4, 'b': 3}))
print(mul.description)
print(mul.args)
```

```python
llm = ChatGroq(model="llama-3.1-8b-instant", temperature=0)
```

```python
llm_with_tools = llm.bind_tools([mul, add])
```

```python
print(llm_with_tools.invoke("Multiply 3 and 4").tool_calls)
```
